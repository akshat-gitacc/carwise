```python
import pandas as pd

df = pd.read_csv("../data/processed_dataset.csv")

numerical_features = [
    'vehicle_age',
    'km_driven',
    'mileage',
    'engine',
    'max_power',
    'seats',
    'selling_price'
]

print("Negative values:")
print((df[numerical_features] < 0).sum())

print("\nZero values:")
print((df[numerical_features] == 0).sum())
```

```python
zero_seat_cars = df[df['seats'] == 0]

print(zero_seat_cars.to_string(index=False))
```

```python
df.loc[df['seats'] == 0, 'seats'] = 5
```

```python
df.to_csv("../data/processed_dataset.csv", index=False)
```
